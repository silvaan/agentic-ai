## Domínio

Descrição do domínio, de quem usaria o sistema e de como se sabe que uma resposta está certa. A tarefa deve exigir mais de um tipo de trabalho, como calcular e redigir, pesquisar e revisar ou implementar e testar, e ter casos em que o resultado de um passo muda o passo seguinte. Também deve ser escrita aqui a hipótese da dupla: qual vantagem a divisão em especialistas deveria trazer, como contexto menor, prompts mais focados ou verificação independente. As ferramentas usadas pelas duas versões são definidas aqui, uma vez só.

## Agente único

Agente criado com `create_agent`, com todas as ferramentas do domínio, o mesmo modelo do sistema multiagente e uma mensagem de sistema que descreva a tarefa inteira. Esta é a linha de base da comparação, e deve ser escrita com o mesmo cuidado que os especialistas.

## Especialistas

Pelo menos dois especialistas criados com `create_agent`, cada um com a própria mensagem de sistema, as próprias ferramentas e um papel que o outro não cumpre. Cada especialista vira uma ferramenta do supervisor, como um subagente, e a docstring da ferramenta diz o que ele recebe e o que devolve.

## Supervisor

Supervisor que decide a cada passo qual especialista trabalha, com `create_agent` e os especialistas como ferramentas ou com um `StateGraph` em que o supervisor devolve `Command(goto=...)`. Pelo menos um caso deve mostrar o supervisor usando um resultado para decidir o passo seguinte, como uma revisão reprovada que volta ao especialista. O limite de passos ou de chamadas fica no código, e a trajetória de um caso deve ser mostrada com `pretty_print`.

## Avaliação

Conjunto de pelo menos dez casos, como uma lista de dicionários, com a entrada e a resposta de referência, incluindo casos difíceis e pelo menos um em que o resultado de um especialista deveria mudar o plano. As duas versões devem rodar todos os casos, registrando para cada um:

- se a resposta está certa, por uma verificação em Python ou por um LLM como juiz com saída estruturada, conferido à mão em uma amostra;
- o número de chamadas ao modelo e de delegações;
- os tokens do sistema inteiro, somados com `get_usage_metadata_callback`, e a latência.

## Comparação

Tabela com uma linha por versão e as médias de acerto, chamadas, tokens e latência, seguida de uma tabela dos casos em que as duas versões discordaram. A trajetória de pelo menos um desses casos deve ser mostrada para cada versão.
